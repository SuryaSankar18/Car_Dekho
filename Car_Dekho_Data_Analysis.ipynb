**Detect Outliers Using IQR**

```python
Q1 = df['Selling_Price'].quantile(0.25)
Q3 = df['Selling_Price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Selling_Price'] < lower_bound) |
    (df['Selling_Price'] > upper_bound)
]

print("Number of outliers:", len(outliers))
print(outliers)
```

```python
def find_outliers(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower) |
        (data[column] > upper)
    ]

    return outliers


print(find_outliers(df, 'Selling_Price'))
```

# **Depreciation Analysis**

```python
df['Depreciation'] = (
    df['Present_Price'] - df['Selling_Price']
)

df[['Present_Price', 'Selling_Price', 'Depreciation']].head()
```

```python
df['Depreciation_Percentage'] = (df['Depreciation'] / df['Present_Price']) * 100

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Depreciation_Percentage',
    kde=True,
    bins=30
)

plt.title("Distribution of Depreciation Percentage")
plt.xlabel("Depreciation (%)")
plt.ylabel("Number of Cars")

plt.show()
```

**Car Age vs Depreciation**

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Car_Age',
    y='Depreciation_Percentage',
    hue='Fuel_Type'
)

plt.title("Car Age vs Depreciation Percentage")
plt.xlabel("Car Age")
plt.ylabel("Depreciation (%)")

plt.show()
```

**Average depreciation by fuel type**

```python
depreciation_fuel = (
    df.groupby('Fuel_Type')['Depreciation_Percentage']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(9, 6))

sns.barplot(
    x=depreciation_fuel.index,
    y=depreciation_fuel.values
)

plt.title("Average Depreciation Percentage by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Depreciation (%)")

plt.show()
```

**Average depreciation by transmission**

```python
depreciation_transmission = (
    df.groupby('Transmission')['Depreciation_Percentage']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 6))

sns.barplot(
    x=depreciation_transmission.index,
    y=depreciation_transmission.values
)

plt.title("Average Depreciation by Transmission")
plt.xlabel("Transmission")
plt.ylabel("Average Depreciation (%)")

plt.show()
```
